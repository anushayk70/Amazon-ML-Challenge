```python
import pandas as pd

datasets = {
    "Source 1": train_s1,
    "Source 2": train_s2,
    "Source 3": train_s3
}

# ============================================================
# 1. WHITESPACE / EMPTY STRING CHECK
# ============================================================

print("\n" + "="*70)
print("EMPTY / WHITESPACE BUSINESS NAMES")
print("="*70)

for name, df in datasets.items():
    s = df["business_name"].dropna().astype(str)

    empty = (s.str.strip() == "").sum()

    print(f"{name}: {empty}")


# ============================================================
# 2. CASE VARIATION
# ============================================================

print("\n" + "="*70)
print("BUSINESS NAMES AFFECTED BY CASE")
print("="*70)

for name, df in datasets.items():
    s = df["business_name"].dropna().astype(str)

    case_duplicates = (
        s.str.lower().nunique()
    )

    original_unique = s.nunique()

    print(f"{name}")
    print("Original unique names:", original_unique)
    print("Lowercase unique names:", case_duplicates)
    print("Potential case-collapsed names:",
          original_unique - case_duplicates)


# ============================================================
# 3. ADDRESS LENGTH
# ============================================================

print("\n" + "="*70)
print("BUSINESS ADDRESS LENGTH")
print("="*70)

for name, df in datasets.items():

    s = df["business_address"].dropna().astype(str)
    lengths = s.str.len()

    print(f"\n{name}")
    print("Minimum:", lengths.min())
    print("Maximum:", lengths.max())
    print("Mean:", round(lengths.mean(), 2))
    print("Median:", lengths.median())
    print("95th percentile:", lengths.quantile(0.95))


# ============================================================
# 4. GROUND TRUTH MATCH COUNTS
# ============================================================

print("\n" + "="*70)
print("GROUND TRUTH MATCH COUNTS")
print("="*70)

gt = ground_truth.copy()

def count_matches(value):
    if pd.isna(value):
        return 0

    value = str(value).strip()

    if value == "":
        return 0

    # Handle common separators used in match lists
    if "|" in value:
        return len([x for x in value.split("|") if x.strip()])

    if "," in value:
        return len([x for x in value.split(",") if x.strip()])

    return 1


gt["match_count"] = gt["matched_entity_ids"].apply(count_matches)

print(gt["match_count"].value_counts().sort_index())


# ============================================================
# 5. GROUND TRUTH SUMMARY
# ============================================================

print("\n" + "="*70)
print("GROUND TRUTH SUMMARY")
print("="*70)

print("Total Source 1 entities:", len(gt))
print("Zero-match entities:", (gt["match_count"] == 0).sum())
print("One-match entities:", (gt["match_count"] == 1).sum())
print("Multiple-match entities:", (gt["match_count"] > 1).sum())
```

```python
import re

def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Replace punctuation/special characters with spaces
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    # Remove punctuation while PRESERVING Unicode letters/numbers
    text = "".join(
        char if char.isalnum() or char.isspace() else " "
        for char in text
    )

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
# Re-create normalized columns using the corrected function

for df in [train_s1, train_s2, train_s3]:
    df["name_normalized"] = df["business_name"].apply(normalize_text)
    df["address_normalized"] = df["business_address"].apply(normalize_text)

print(train_s2[[
    "business_name",
    "name_normalized"
]].head(10).to_string(index=False))
```

```python
# STEP 7 — Candidate Generation

# Create dictionaries mapping each blocking key
# to the rows belonging to that block.

s2_blocks = (
    train_s2.groupby("name_block")["entity_id"]
    .apply(list)
    .to_dict()
)

s3_blocks = (
    train_s3.groupby("name_block")["entity_id"]
    .apply(list)
    .to_dict()
)

print("Source 2 blocks:", len(s2_blocks))
print("Source 3 blocks:", len(s3_blocks))


# Generate candidates for Source 1
candidates = []

for _, row in train_s1.iterrows():

    source1_id = row["entity_id"]
    block = row["name_block"]

    # Candidates from Source 2
    for entity_id in s2_blocks.get(block, []):
        candidates.append(
            (source1_id, entity_id, "S2")
        )

    # Candidates from Source 3
    for entity_id in s3_blocks.get(block, []):
        candidates.append(
            (source1_id, entity_id, "S3")
        )


candidates_df = pd.DataFrame(
    candidates,
    columns=[
        "source1_entity_id",
        "candidate_entity_id",
        "candidate_source"
    ]
)

print("\nTotal candidate pairs:", len(candidates_df))

print("\nCandidates by source:")
print(candidates_df["candidate_source"].value_counts())

print("\nSample candidates:")
print(candidates_df.head(10))
```

```python
import pandas as pd
import zipfile

# ============================================================
# STEP 7 — LOAD DATA + CREATE BLOCKS + GENERATE CANDIDATES
# ============================================================

zip_path = "/content/drive/MyDrive/Amazon-ML-Challenge/6ab10eb3b23ba_student_resource.zip"

# ---------- 1. Load datasets ----------
with zipfile.ZipFile(zip_path, "r") as z:

    train_s1 = pd.read_csv(
        z.open("student_resource/dataset/train/train_source1.tsv"),
        sep="\t"
    )

    train_s2 = pd.read_csv(
        z.open("student_resource/dataset/train/train_source2.tsv"),
        sep="\t"
    )

    train_s3 = pd.read_csv(
        z.open("student_resource/dataset/train/train_source3.tsv"),
        sep="\t"
    )

print("Data loaded:")
print("Source 1:", train_s1.shape)
print("Source 2:", train_s2.shape)
print("Source 3:", train_s3.shape)


# ---------- 2. Create fast blocking keys ----------
def create_name_block(df):
    return (
        df["business_name"]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.replace(r"[^a-z0-9]", "", regex=True)
        .str[:10]
    )


train_s1["name_block"] = create_name_block(train_s1)
train_s2["name_block"] = create_name_block(train_s2)
train_s3["name_block"] = create_name_block(train_s3)

print("\nBlocking keys created.")

print("Source 1 blocks:", train_s1["name_block"].nunique())
print("Source 2 blocks:", train_s2["name_block"].nunique())
print("Source 3 blocks:", train_s3["name_block"].nunique())


# ---------- 3. Generate candidates using merges ----------
s1_blocks = train_s1[
    ["entity_id", "name_block"]
].rename(
    columns={"entity_id": "source1_entity_id"}
)

s2_blocks = train_s2[
    ["entity_id", "name_block"]
].rename(
    columns={"entity_id": "candidate_entity_id"}
)

s3_blocks = train_s3[
    ["entity_id", "name_block"]
].rename(
    columns={"entity_id": "candidate_entity_id"}
)


print("\nGenerating Source 2 candidates...")

candidates_s2 = s1_blocks.merge(
    s2_blocks,
    on="name_block",
    how="inner"
)

candidates_s2["candidate_source"] = "S2"


print("Generating Source 3 candidates...")

candidates_s3 = s1_blocks.merge(
    s3_blocks,
    on="name_block",
    how="inner"
)

candidates_s3["candidate_source"] = "S3"


# ---------- 4. Combine ----------
candidates_df = pd.concat(
    [candidates_s2, candidates_s3],
    ignore_index=True
)

candidates_df = candidates_df[
    [
        "source1_entity_id",
        "candidate_entity_id",
        "candidate_source"
    ]
]


print("\n==================================================")
print("CANDIDATE GENERATION COMPLETE")
print("==================================================")

print("Total candidate pairs:", len(candidates_df))

print("\nCandidates by source:")
print(candidates_df["candidate_source"].value_counts())

print("\nSample:")
print(candidates_df.head(10))
```

```python
# ============================================================
# RECOVER DATASET + LOAD SOURCE FILES
# ============================================================

import os
import glob
import zipfile
import pandas as pd
import re
import unicodedata

print("=" * 70)
print("RECOVERING AMAZON ML CHALLENGE DATA")
print("=" * 70)

# ------------------------------------------------------------
# 1. Google Drive
# ------------------------------------------------------------

from google.colab import drive

drive.mount("/content/drive", force_remount=True)

# ------------------------------------------------------------
# 2. Locate the student resource ZIP
# ------------------------------------------------------------

zip_path = "/content/drive/MyDrive/Amazon-ML-Challenge/6ab10eb3b23ba_student_resource.zip"

if not os.path.exists(zip_path):
    raise FileNotFoundError(
        f"ZIP file not found:\n{zip_path}"
    )

print("\nZIP found:")
print(zip_path)

# ------------------------------------------------------------
# 3. Inspect ZIP contents
# ------------------------------------------------------------

print("\nSearching inside ZIP...")

with zipfile.ZipFile(zip_path, "r") as z:
    all_files = z.namelist()

print(f"Total files inside ZIP: {len(all_files)}")

# Show TSV files
tsv_files = [
    f for f in all_files
    if f.lower().endswith((".tsv", ".csv"))
]

print("\nDataset files found:")
for f in tsv_files:
    print(" ->", f)

# ------------------------------------------------------------
# 4. Find Source 1, Source 2, Source 3 and Ground Truth
# ------------------------------------------------------------

def find_file(patterns):
    """
    Find a file inside the ZIP using multiple possible patterns.
    """
    for f in tsv_files:
        name = os.path.basename(f).lower()

        for pattern in patterns:
            if pattern in name:
                return f

    return None


source1_file = find_file([
    "train_source1",
    "source1"
])

source2_file = find_file([
    "train_source2",
    "source2"
])

source3_file = find_file([
    "train_source3",
    "source3"
])

ground_truth_file = find_file([
    "train_ground_truth",
    "ground_truth"
])

print("\nDetected files:")
print("Source 1      :", source1_file)
print("Source 2      :", source2_file)
print("Source 3      :", source3_file)
print("Ground Truth  :", ground_truth_file)

# ------------------------------------------------------------
# 5. Make sure required files were found
# ------------------------------------------------------------

if source1_file is None:
    raise FileNotFoundError(
        "Could not find Source 1 inside the ZIP."
    )

if source2_file is None:
    raise FileNotFoundError(
        "Could not find Source 2 inside the ZIP."
    )

if source3_file is None:
    raise FileNotFoundError(
        "Could not find Source 3 inside the ZIP."
    )

# ------------------------------------------------------------
# 6. Extract the required files
# ------------------------------------------------------------

extract_dir = "/content/amazon_ml_data"

os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as z:

    files_to_extract = [
        source1_file,
        source2_file,
        source3_file
    ]

    if ground_truth_file is not None:
        files_to_extract.append(ground_truth_file)

    for file_name in files_to_extract:
        z.extract(file_name, extract_dir)

print("\nFiles extracted successfully.")

# ------------------------------------------------------------
# 7. Get actual extracted paths
# ------------------------------------------------------------

def locate_extracted_file(filename):
    basename = os.path.basename(filename)

    matches = glob.glob(
        os.path.join(extract_dir, "**", basename),
        recursive=True
    )

    if not matches:
        raise FileNotFoundError(
            f"Could not locate extracted file: {basename}"
        )

    return matches[0]


source1_path = locate_extracted_file(source1_file)
source2_path = locate_extracted_file(source2_file)
source3_path = locate_extracted_file(source3_file)

print("\nActual paths:")
print("Source 1:", source1_path)
print("Source 2:", source2_path)
print("Source 3:", source3_path)

# ------------------------------------------------------------
# 8. Load datasets
# ------------------------------------------------------------

print("\nLoading Source 1...")
train_s1 = pd.read_csv(
    source1_path,
    sep="\t",
    low_memory=False
)

print("Source 1 shape:", train_s1.shape)

print("\nLoading Source 2...")
train_s2 = pd.read_csv(
    source2_path,
    sep="\t",
    low_memory=False
)

print("Source 2 shape:", train_s2.shape)

print("\nLoading Source 3...")
train_s3 = pd.read_csv(
    source3_path,
    sep="\t",
    low_memory=False
)

print("Source 3 shape:", train_s3.shape)

# ------------------------------------------------------------
# 9. Display columns
# ------------------------------------------------------------

print("\nSource 1 columns:")
print(train_s1.columns.tolist())

print("\nSource 2 columns:")
print(train_s2.columns.tolist())

print("\nSource 3 columns:")
print(train_s3.columns.tolist())

print("\n" + "=" * 70)
print("DATA LOADING COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 7 — STRONG BLOCKING + CANDIDATE ESTIMATION
# ============================================================

import pandas as pd
import re
import unicodedata

print("=" * 70)
print("STEP 7 — STRONG BLOCKING + CANDIDATE ESTIMATION")
print("=" * 70)


# ============================================================
# STEP 7A — TEXT NORMALIZATION
# ============================================================

print("\n[1/6] Creating normalized business names...")


def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Keep letters, numbers and spaces
    text = "".join(
        char if char.isalnum() or char.isspace() else " "
        for char in text
    )

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


for df in [train_s1, train_s2, train_s3]:

    if "name_normalized" not in df.columns:
        df["name_normalized"] = (
            df["business_name"]
            .fillna("")
            .astype(str)
            .apply(normalize_text)
        )


print("✓ Name normalization complete")


# ============================================================
# STEP 7B — CREATE STRONG BLOCKING KEY
# ============================================================

print("\n[2/6] Creating strong blocking keys...")


def create_strong_block(name):

    if pd.isna(name):
        return ""

    name = str(name).strip()

    if not name:
        return ""

    # Split into words
    words = name.split()

    # First word
    first_word = words[0]

    # First character of first word
    first_char = first_word[0] if first_word else ""

    # Length bucket
    length_bucket = min(len(name) // 5, 10)

    return f"{first_char}_{length_bucket}"


for df in [train_s1, train_s2, train_s3]:

    df["strong_block"] = (
        df["name_normalized"]
        .apply(create_strong_block)
    )


print("✓ Strong blocking keys created")


# ============================================================
# STEP 7C — BLOCK STATISTICS
# ============================================================

print("\n[3/6] Inspecting block statistics...")

s1_block_sizes = train_s1["strong_block"].value_counts()
s2_block_sizes = train_s2["strong_block"].value_counts()
s3_block_sizes = train_s3["strong_block"].value_counts()

print("\nSource 1 unique blocks:", len(s1_block_sizes))
print("Source 2 unique blocks:", len(s2_block_sizes))
print("Source 3 unique blocks:", len(s3_block_sizes))

print("\nLargest Source 1 blocks:")
print(s1_block_sizes.head(10))

print("\nLargest Source 2 blocks:")
print(s2_block_sizes.head(10))

print("\nLargest Source 3 blocks:")
print(s3_block_sizes.head(10))


# ============================================================
# STEP 7D — FIND COMMON BLOCKS
# ============================================================

print("\n[4/6] Finding common blocks between Source 1 and Source 2...")

common_blocks = (
    s1_block_sizes.index
    .intersection(s2_block_sizes.index)
)

print("Common blocks:", len(common_blocks))


# ============================================================
# STEP 7E — ESTIMATE CANDIDATE PAIRS
# ============================================================

print("\n[5/6] Calculating candidate count...")

s1_common = s1_block_sizes.loc[common_blocks]
s2_common = s2_block_sizes.loc[common_blocks]

candidate_counts = s1_common * s2_common

total_candidates = candidate_counts.sum()

print("\n" + "=" * 60)
print("STRONG BLOCK CANDIDATE ESTIMATE")
print("=" * 60)

print("Source 1 records:", len(train_s1))
print("Source 2 records:", len(train_s2))
print("Common blocks:", len(common_blocks))
print(
    "Estimated S1-S2 candidate pairs:",
    int(total_candidates)
)


# ============================================================
# STEP 7F — IDENTIFY LARGE BLOCKS
# ============================================================

print("\n[6/6] Inspecting potentially large blocks...")

block_summary = pd.DataFrame({
    "s1_count": s1_common,
    "s2_count": s2_common
})

block_summary["candidate_pairs"] = (
    block_summary["s1_count"]
    * block_summary["s2_count"]
)

block_summary = block_summary.sort_values(
    "candidate_pairs",
    ascending=False
)

print("\nTop 20 blocks by candidate-pair count:")

print(
    block_summary.head(20).to_string()
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STEP 7 COMPLETE")
print("=" * 70)

print("\nSource 1 unique blocks:", len(s1_block_sizes))
print("Source 2 unique blocks:", len(s2_block_sizes))
print("Source 3 unique blocks:", len(s3_block_sizes))
print("Common S1-S2 blocks:", len(common_blocks))
print(
    "Estimated S1-S2 candidates:",
    f"{int(total_candidates):,}"
)

print("\nColumns now available in each source:")

print(train_s1.columns.tolist())

print("\n✓ train_s1 ready")
print("✓ train_s2 ready")
print("✓ train_s3 ready")
print("✓ Blocking analysis ready")
```

# STEP 8 — EDA CONCLUSIONS

## 8.1 Dataset Overview

The training data consists of three source datasets and a ground-truth mapping:

- Source 1: 2,206,821 records
- Source 2: 5,034,616 records
- Source 3: 5,285,603 records
- Ground Truth: 2,206,821 records

All three source datasets contain the following attributes:

- `entity_id`
- `business_name`
- `business_address`
- `country`

The ground-truth dataset contains:

- `source1_entity_id`
- `matched_entity_ids`

---

## 8.2 Data Preparation

Business names were normalized before blocking.

The normalization process included:

- Converting text to lowercase
- Unicode normalization
- Removing punctuation while preserving letters and numbers
- Normalizing whitespace
- Handling missing values

A normalized business-name representation was created as `name_normalized`.

---

## 8.3 Blocking Analysis

A strong blocking key was created using the normalized business name.

The blocking analysis produced:

- Source 1 unique blocks: 375
- Source 2 unique blocks: 2,624
- Source 3 unique blocks: 2,596
- Common Source 1–Source 2 blocks: 375

Blocking is necessary because comparing every record in Source 1 with every record in Source 2 would create an extremely large comparison space.

---

## 8.4 Candidate-Pair Estimation

The current blocking strategy produces an estimated:

**93,205,788,932 S1–S2 candidate pairs**

or approximately:

**93.2 billion candidate pairs.**

The largest blocks produce billions of potential comparisons individually. For example, the `s_5` block produces approximately 6.29 billion S1–S2 candidate pairs.

Therefore, the current blocking strategy is not selective enough for direct candidate-pair generation.

---

## 8.5 Key EDA Findings

1. The three source datasets contain millions of business records, making exhaustive pairwise comparison computationally impractical.

2. Business names require normalization before similarity comparison because differences in case, punctuation, Unicode representation and whitespace can affect matching.

3. Blocking can substantially organize the search space by grouping potentially related records.

4. The current strong-blocking strategy still produces an extremely large estimated candidate space of approximately 93.2 billion S1–S2 pairs.

5. Direct materialization of all estimated candidate pairs should therefore be avoided.

6. A more selective or multi-stage blocking strategy is required before candidate generation.

---

## 8.6 Recommendations for the Next Stage

Based on the EDA and blocking analysis, the next stage should focus on:

- Refining the blocking strategy
- Using multiple blocking keys
- Reducing oversized blocks
- Generating manageable candidate sets
- Engineering similarity features for candidate pairs
- Training and evaluating entity-resolution models

These steps will be handled by the candidate-generation and machine-learning stages of the project.
